```python
import random
from datetime import datetime, timedelta

# Cấu hình dữ liệu mẫu tiếng Việt
ho = ["Nguyễn", "Trần", "Lê", "Phạm", "Hoàng", "Phan", "Vũ", "Đặng", "Bùi", "Đỗ"]
dem = ["Văn", "Thị", "Anh", "Minh", "Quang", "Hữu", "Đình", "Thành", "Ngọc", "Thu"]
ten = ["Anh", "Bình", "Chi", "Dũng", "Em", "An", "Giang", "Hương", "Khánh", "Linh", "Trang", "Nam", "Sơn"]
tinh = ["Hà Nội", "TP.HCM", "Đà Nẵng", "Nam Định", "Hải Phòng", "Cần Thơ", "Huế"]
dv_names = ["Cloud Server", "Dedicated Server", "Web Hosting", "Email Business", "VPS Pro", "CDN Service", "SSL Certificate"]

def generate_sql():
    sql_lines = ["USE ContractManagement;\n", "SET FOREIGN_KEY_CHECKS = 0;\n"]
    
    # 1. Sinh 510 Services
    services = []
    for i in range(1, 511):
        s_id = f"S{i:03d}"
        s_name = f"{random.choice(dv_names)} {random.choice(['Basic', 'Gold', 'Platinum', 'VIP'])}"
        price = random.randint(500000, 20000000)
        services.append((s_id, price))
        sql_lines.append(f"INSERT INTO Services VALUES ('{s_id}', '{s_name}', 'Description for {s_name}', {price});\n")

    # 2. Sinh 510 Customers
    cust_ids = []
    for i in range(1, 511):
        c_id = f"C{i:03d}"
        c_name = f"{random.choice(ho)} {random.choice(dem)} {random.choice(ten)}"
        addr = f"{random.randint(1, 999)} Đường {random.choice(ten)}, {random.choice(tinh)}"
        phone = f"09{random.randint(10000000, 99999999)}"
        email = f"user{i}@gmail.com"
        cust_ids.append(c_id)
        sql_lines.append(f"INSERT INTO Customers VALUES ('{c_id}', '{c_name}', '{addr}', '{phone}', '{email}');\n")

    # 3. Sinh 510 Contracts
    contracts = []
    for i in range(1, 511):
        ct_id = f"CT{i:03d}"
        c_id = cust_ids[i-1]
        s_id, unit_price = random.choice(services)
        qty = random.randint(1, 5)
        total = unit_price * qty
        sign_date = datetime(2023, 1, 1) + timedelta(days=random.randint(0, 1100))
        duration = random.randint(6, 24)
        status = random.choice(['Active', 'Expired', 'Terminated'])
        contracts.append((ct_id, total, sign_date))
        sql_lines.append(f"INSERT INTO Contracts VALUES ('{ct_id}', '{c_id}', '{s_id}', '{sign_date.date()}', {duration}, {total}, '{status}', {qty});\n")

    # 4. Sinh 510 Invoices (Dựa trên Contracts)
    for i in range(1, 511):
        inv_id = f"INV{i:03d}"
        ct_id, total, sign_date = contracts[i-1]
        issue_date = sign_date + timedelta(days=random.randint(1, 5))
        status = random.choice(['Paid', 'Unpaid'])
        sql_lines.append(f"INSERT INTO Invoices VALUES ('{inv_id}', '{ct_id}', '{issue_date.date()}', {total}, 'Monthly bill', '{status}');\n")

    # 5. Sinh 510 Payments (Logic: Amount <= TotalValue)
    for i in range(1, 511):
        pay_id = f"P{i:03d}"
        ct_id, total, sign_date = contracts[i-1]
        pay_date = sign_date + timedelta(days=random.randint(5, 15))
        amount = total if random.random() > 0.2 else total * 0.5 # 80% trả hết, 20% trả nửa
        method = random.choice(['Bank Transfer', 'Cash', 'Credit Card'])
        sql_lines.append(f"INSERT INTO Payments VALUES ('{pay_id}', '{ct_id}', '{pay_date.date()}', {amount}, '{method}');\n")

    sql_lines.append("SET FOREIGN_KEY_CHECKS = 1;\n")
    
    with open("insert_data.sql", "w", encoding="utf-8") as f:
        f.writelines(sql_lines)
    print("Check file insert_data.sql !")

generate_sql()
```
